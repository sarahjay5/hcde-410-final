## 1. Overview & Research Questions / Hypotheses

### Overview
This project investigates what is associated with how long dogs stay at the Austin Animal Center (AAC) before leaving alive (adopted, returned to owner, or transferred). Length of stay (LOS) matters to the dogs, who experience kennel stress while they wait, and to the shelter, which has limited space and staff. After feedback on A4, I narrowed the project from four sub-questions to two priorities: **age** and **coat color** once intake condition is accounted for. Intake type and season become secondary, exploratory analyses.

### Main research question
**What factors are associated with how long a dog stays at the Austin Animal Center before a live outcome?**

### Sub-questions and hypotheses

**RQ1 (priority): How does LOS differ by age group at intake?**
- *H1:* Older dogs have a longer median LOS than younger dogs, and puppies have the shortest. This holds after adjusting for intake condition, intake type, and year.

**RQ2 (priority): Do black dogs stay longer than non-black dogs once intake condition (and age) is taken into account?**
- *H2:* Black dogs do not have a meaningfully longer LOS than non-black dogs after adjustment. I will treat a difference of under about 10% in median LOS as practically negligible (this threshold is my own judgment call, and I will report the full confidence interval so readers can judge for themselves).
- *H3:* The black-dog/non-black-dog difference, if any, is not the same across intake conditions. Sick or injured dogs stay longer overall, and this may confound a naive color comparison.

**RQ3 (secondary, exploratory): Does LOS differ by intake type or by month/season of intake?**
- No strong directional hypothesis. I will report these as descriptive patterns and will not draw causal conclusions.

### Scope
Findings describe association at one shelter in one time period (Oct 2013 to May 2025). They are not reasons or causes, and they are not claims about shelters in general.

## 2. Background / Related Work

**The "black dog syndrome" belief.** The idea that black dogs wait longer for adoption is widespread in popular media. A 2008 TIME article describes the belief and offers explanations such as superstition, black dogs photographing poorly for shelter websites and flyers, and dark dogs being hard to see in dimly lit kennels. It also mentions a Pethealth analysis of 679 shelters that found black dogs stayed only about two days longer on average, and one Austin-area shelter that reported a large share of its animals were black [1]. This is mostly anecdotal and journalistic evidence, which is a reason to test the claim with data rather than assume it.

**What research has found.** Findings are mixed:
- Svoboda & Hoffman analyzed four years of intake and outcome records for roughly 16,700 dogs at two Pacific Northwest shelters. Black dogs did not have significantly longer time to adoption or higher euthanasia rates. Age and breed group were more consistent predictors, and older dogs had longer stays [2]. Their sample excluded very young dogs and dogs over 13 years old.
- Brown, Davidson & Zuefle studied two no-kill shelters in New York State. Puppies had the shortest LOS, LOS rose steadily with age, and coat color and sex had no detectable effect. The authors warn that color and breed effects may be local and may not generalize across shelters [3].
- Protopopova et al., as summarized in [2], also reported that LOS was not significantly associated with color, age, or sex.
-Patronek & Crowe studied two years of intake records (about 21,400 dog intakes) at the Pima Animal Care Center, a large open-admission municipal shelter in Tucson, Arizona. Live release was the same for black and non-black dogs, and coat color had no effect after adjusting for other factors, including age, sex, and condition at intake [4]. They also describe a study of a Louisville, KY shelter (Sinski et al., 2016) in which black dogs were about 24% less likely to be adopted in a simple analysis, though the pattern interacted with size and breed, so the results were inconclusive [4].

**How this shapes my study.**
1. *Age first.* Because age is the most consistent predictor in prior work [2, 3], it is a priority question and a required control in every model.
2. *Color is tested, not assumed.* Since the literature is split, I treat black dog syndrome as an open question for Austin rather than a given.
3. *Adjusting for condition.* Patronek & Crowe's use of condition at intake as a covariate [4], plus my A4 feedback, supports comparing black and non-black dogs within similar intake conditions.
4. Local data matters. Brown et al. [3] argue that shelters should look at their own data, which is a direct justification for focusing on Austin. The Austin Animal Center also publishes more than ten years of intake and outcome records as open data, which makes a study of its own length-of-stay patterns possible.
5. *Design relevance.* The photo and visibility explanations in [1] connect to my human-centered goal: if color has any effect, it may work through how dogs are presented on listing pages, which shelters can change.

## 3. Methodology

### 3.1 Data, license, and sources

The Attribution section asks users to credit both the City of Austin and the source department (here, Austin Animal Services / Austin Animal Center). The Confidentiality section says users must not re-identify the data or combine it with other datasets to re-identify individuals, which supports my decision not to use or publish found-location. The terms also disclaim any warranty on accuracy and say the City cannot vouch for analyses done with the data, so I will present results as my own analysis.

City of Austin, Texas Open Data Portal. Retrieved November 15, 2026, from https://data.austintexas.gov/d/9t4d-g238.
- **Why this pair:** Austin switched shelter database systems on May 5, 2025, and newer records live in separate datasets. Using only the older pair keeps the records under one system and one set of field definitions. I will download the files early and save a dated copy so the analysis is reproducible.

### 3.2 Building the analysis table
1. **Filter to dogs** (`Animal Type == "Dog"`). Drop intake types that are not about rehoming, such as owner-requested euthanasia, and note how many rows this removes.
2. **Match intakes to outcomes.** Sort both files by date and use `pd.merge_asof(..., by="Animal ID", direction="forward")` so each intake is paired with that animal's next outcome on or after the intake date. This handles dogs that came in more than once. I will check a sample of repeat-intake dogs by hand to confirm the matching.
3. **Compute LOS** as days from intake date to outcome date. I will report how many stays are 0 days and keep them (same-day returns to owner are real), but also run results excluding them as a sensitivity check.
4. **Define the outcome.** The main analysis uses *live outcomes only*: Adoption, Return to Owner, and Transfer. Dogs that died, were euthanized, or were otherwise lost have no "left alive" stay, so they are excluded from the main LOS comparison and counted and reported separately by group. Intakes with no matching outcome (dogs still in the shelter when the data ends) are also excluded from the main model and counted.
5. **Age.** Parse the text field ("2 years", "3 weeks") into days with a small tested function, then group into four bands: under 6 months, 6 months to under 2 years, 2 to under 7 years, and 7 years or older. Rows with unparseable ages get their own category or are dropped, and I will report how many.
6. **Black dog rule.** Colors are combinations such as "Black/White." My written rule: split the color field on "/" and look at the first listed color. *Black-primary* means the first color is Black (including Black Smoke or Black Brindle variants, which I will list explicitly). *Partly black* means Black appears only as a second color. *Non-black* means no Black. The main comparison is black-primary vs. non-black. A sensitivity analysis uses "any black" vs. "no black," and another uses strictly solid "Black."
7. **Intake condition.** Group into Normal, Sick/Injured/Medical-related, and Other, after looking at the actual category counts.

### 3.3 Analysis, and why these methods
- **Descriptive statistics and plots (all RQs).** LOS is strongly right-skewed (most dogs leave quickly, a few stay very long), so I will report medians and interquartile ranges and use box/violin plots on a log scale rather than relying on means.
- **Group comparisons.** Kruskal-Wallis and Mann-Whitney tests are appropriate because LOS is not normally distributed. I will report effect sizes and bootstrap confidence intervals for the ratio of median LOS, not just p-values, because with tens of thousands of dogs even trivial differences can be "significant."
- **Adjusted model (RQ1, RQ2).** A regression on log(LOS + 1) with age band, black-primary indicator, intake condition, intake type, and intake year. The coefficients are interpretable as percent differences in LOS, which matches how I want to talk about results. I will add a black × condition interaction to test H3. This directly applies the A4 feedback that sick dogs may stay longer because of condition, not color or age.
- **Survival-analysis check.** Because some dogs are still in the shelter at the data cutoff, and because dogs leave by different routes, I will run Kaplan-Meier curves and a Cox proportional hazards model as a robustness check, treating non-live outcomes and still-in-shelter dogs as censored. If the survival and regression results agree, I can be more confident the findings do not depend on how I handled excluded dogs.
- **Comparison to prior work.** I will rerun the color comparison on dogs aged 1 to 13 years, matching the age range in [2], so my results can be compared with theirs directly.
- **Secondary (RQ3).** LOS by intake type and by month of intake, shown descriptively, with a clear note that these are exploratory and not adjusted for every confounder. Multiple comparisons will be noted rather than hidden.

### 3.4 Changes at the shelter over time
Shelter policies and operations changed over the 11+ years, including during 2020, and the data does not explain why. Intake year is a covariate in the models, and I will check whether results differ before and after 2020. I will not interpret year effects as causes.

### 3.5 How findings will be presented
A final report with a small number of clear figures: LOS by age band, black vs. non-black LOS within each intake condition (with confidence intervals), and survival curves. Each figure will be labeled to avoid overclaiming. The notebook will run top to bottom from the downloaded files, with fixed random seeds and the cleaning rules written out, so someone else can reproduce it.

### 3.6 Ethical and human-centered considerations
- **Who is affected.** The records describe animals, but shelter staff and adopters are the people who would use the results. The knowledge serves shelters and listing-site designers deciding where to put extra visibility (better photos, featured listings, fee promotions) and when to hold events.
- **Privacy.** I will not use or publish the found-location field, since it could point to neighborhoods or individuals.
- **Breed.** Breed labels at shelters are usually visual guesses, so I will leave breed out of the models or use it only descriptively, and I will not label any breed as unadoptable.
- **Fair representation.** Only one shelter and one city are represented, with its own intake policies. Results should not be generalized without further research.
- **Responsible framing.** I will describe findings as associations. If a group of dogs stays longer, the takeaway should be "this group may need more visibility," never "this group is less worth saving." If black dog syndrome does not show up in Austin, shelters can spend promotion effort on dogs that do wait longer. If it does, the photo and listing-page explanations are worth testing.

### 3.7 Limitations I expect
Color and age are estimated by staff, not measured. Dog behavior and size are not in the data. A dog's LOS can reflect shelter decisions, such as holds or transfer scheduling, and not just adopter preference. The live-outcome restriction means I cannot say anything about dogs that were euthanized.
