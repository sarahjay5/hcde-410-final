## What Affects How Long Shelter Dogs Stay?

**Course:** HCDE 410  |  **By:** Sarah Jayagaran

## 1. Motivation & Problem

Shelters have limited space and staff, and a dog's outcome (adopted, returned to owner, transferred) can depend on more than its behavior. Age, coat color, how it arrived, its condition, and even the time of year may all play a role.

**Main question:** What factors are associated with how long a dog stays at the Austin Animal Center before leaving alive?

**Sub-questions**
- How does length of stay differ by age group?
- Does it differ by coat color (the "black dog syndrome" idea) once intake condition is taken into account?
- Does it differ by intake type or by month/season?

**Why it matters:** Shelters and adoption websites could use these patterns to decide which dogs need extra visibility (better photos, featured listings, fee promotions) and when to run adoption events. It also informs the design of listing pages, since people choose pets by filtering on exactly these attributes. I hope to learn which factors matter most and which common beliefs the data does not support.

## 2. Data

| Dataset | What it contains | Link |
|---|---|---|
| Austin Animal Center Intakes (10/01/2013 to 05/05/2025) | One row per intake: animal ID, date, intake type, intake condition, animal type, sex, age, breed, color, found location | https://data.austintexas.gov/d/wter-evkm |
| Austin Animal Center Outcomes (10/01/2013 to 05/05/2025) | One row per outcome: animal ID, date, outcome type, age, breed, color | https://data.austintexas.gov/d/9t4d-g238 |

**License / terms:** Both are published on the City of Austin open data portal and are public. Data.gov lists them as non-federal datasets with no specific license attached, so the portal's terms of use apply. I will cite the City of Austin as the source.

**Why they fit:** They are free, cover more than ten years and tens of thousands of dogs, and are built to be joined on animal ID. I will filter to dogs and join each intake to its next outcome to compute length of stay (days between the two dates). I am using this older pair because Austin changed shelter database systems on May 5, 2025, and the newer records are published as separate datasets.

**Ethical considerations**
- The records describe animals, not people, but found-location fields could point to neighborhoods or individuals, so I will not use or publish them.
- Breed labels are usually visual guesses, so I will treat breed as unreliable or leave it out, and avoid stigmatizing any breed. (The data has no size field, so I will focus on age rather than size.)
- Results come from one shelter with its own policies and should not be generalized to all shelters.
- I will describe findings as associations, not reasons, and avoid implying any group of dogs is less worth saving.

## 3. Unknowns & Dependencies

- Messy data: Some columns, like age and color, are inconsistent, so cleaning may take a while.
- Matching records: Some dogs came to the shelter more than once, so it may be hard to match each visit to the right outcome.
- Changes at the shelter: Shelter rules and systems changed over time (including a new database in 2025), and the data doesn’t explain why.
- Data access: The website could change or be slow, so I’ll download the files early.
